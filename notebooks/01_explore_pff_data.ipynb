# PFFNFL Data Exploration

This notebook demonstrates how to use the data collected by the PFFNFL reverse-engineering project.

It is primarily useful for:
- Understanding public PFF data structure
- Building features for prediction / backtesting models
- Combining with nflverse and other public sources

```python
import pandas as pd
import json
from pathlib import Path

DATA_DIR = Path("../data/processed")

def load_parquet(name):
    path = DATA_DIR / f"{name}.parquet"
    if path.exists():
        return pd.read_parquet(path)
    return pd.DataFrame()
```

```python
# Load collected data
teams = load_parquet("teams")
rosters = load_parquet("rosters")
schedules = load_parquet("schedules")

print("Teams:", len(teams))
print("Rosters:", len(rosters))
print("Schedules:", len(schedules))
```
